```python
# Célula 1: Importar as Bibliotecas Principais
# Uma biblioteca é um conjutno de ferramentas que outras pessoas criaram
# A gente importa e usa! Simples assim.

import pandas as pd          # pd = apelido para pandas (trabalhar com dados)
import numpy as np           # np = apelido para numpy (cálculos matemáticos)
import matplotlib.pyplot as plt # plt = apelido para matplotlib (responsável pelos gráficos)
import seaborn as sns        # sns = apelido para seaborn (gráficos mais bonitos)

print("[x] Bibliotecas importadas com sucesso!")
```

```python
# Célula 2: Carregar o arquivo CSV
# pd.read_csv() = função que lê arquivos CSV

df = pd.read_csv('../dados/ai_job_market_dataset.csv')

print("[x] Dataset carregado!")
print(f"O dataset tem {df.shape[0]} linhas e {df.shape[1]} colunas")
```

```python
# Célula 3: Ver as primieras linhas do dataset
# df.head() = mostra as 5 primeiras linhas

print("As 5 primeiras linhas do dataset:\n")
print(df.head())

print("\n" + "="*80 + "\n")

# Você pode pedir mais linhas:
print("As 10 primeiras linhas:\n")
print(df.head(10))
```

```python
#Célula 4: Informações sobre o dataset
#Quais são as colunas que tem? Qual é o tipo de cada uma?

print("FORMA DO DATASET:")
print(f"Linhas: {df.shape[0]}")
print(f"Colunas: {df.shape[1]}")

print("\n" + "="*80 + "\n")

print("NOME DAS COLUNAS:")
print(df.columns.tolist())

print("\n" + "="*80 + "\n")

print("TIPO DE CADA COLUNA:")
print(df.info())
 
```

```python
#Coluna 5: Tem valores faltando (NaN)?
# NaN = Not a Number = célula vazia

print("VALROES FALTANDO POR COLUNA:")
faltando = df.isnull().sum()
print(faltando)

print("\n" + "="*80 + "\n")

print("TOTAL DE VALORES FALTANDO:")
print(f"Total: {df.isnull().sum().sum()}")

print("\n" + "="*80 + "\n")

print("PERCENTUAL DE DADOS COMPLETOS:")
percentual = (1 - (df.isnull().sum().sum() / (df.shape[0] * df.shape[1]))) * 100
print(f"{percentual:.2f}%")

```

```python
#Célula 6: Resumo estatístico das colunas numéricas
# describe() = mostra média, mín, máx, e etc..

# Aqui é importante prestar atenção no que realmente importa! E no significado!
# count é a contagem de quantos salários existem no dataset;
# mean é a média, que se dá através da soma total dividido pelo número total de cadastros
# std é o desvio padrão - mostrando como os salários variam
# SE a média é 94.722 e o desvio padrão é 58,925, a maioria dos salários está entre ($94.722 - $58.925) $35.797 Mínimo, e ($94.722 + $58.925) $153.647 Máximo
# min é o menor salário entre os 2000 cadastrados.
# 25% significa dizer que 1/4 das pessoas ganham menos de $44.920
# 50% a MEDIANA (meio), significa que metade das pessoas ganham menos de $77.489 e que outra metade ganha MAIS de $77.489
# 75% (Ou seja, 3/4) das pessoas, ganham menos de $143.705 - ou ainda, que 25% ganham MAIS de $143.705, que neste caso seriam os melhores remunerados.
# max é o maior salário, que é a singela quantia de $233.977

print("ESTATÍSTICAS DAS COLUNAS NUMÉRICAS:")
print(df.describe())

print("\n" + "="*80 + "\n")

# E se quiser ver mais detalhes:
print("ESTATÍSTICAS COM MAIS DETALHES:")
print(df.describe(include='all'))
```

```python
#Célula 7: Quantas categorias diferentes tem?
# Isso é importante para entneder o que vocÊ tem!

print("COLUNAS CATEGÓRICAS (texto, não número):")

#Pega apenas as colunas que são texto (object)
colunas_categoricas = df.select_dtypes(include=['object','string']).columns

for coluna in colunas_categoricas:
    print(f"\n{coluna}:")
    print(f" Valores únicos: {df[coluna].nunique()}")
    print(f" Exemplos: {df[coluna].unique()[:5]}")    # Mostra os 5 primeiros 
```

```python
#Célula 8: Fazer o primeiro gráfico
#Vamos contar quantos registros tem de cada valor em uma coluna

#Escolha uma coluna categórica
#Tente descobrir o nome exato de uma coluna olhando para df.columns

#Exemplo:
coluna_para_grafico = df.columns[0]   # Pega a primeira coluna

print(f"Contando valores em: {coluna_para_grafico}\n")
contagens = df[coluna_para_grafico].value_counts()
print(contagens)

print("\n" + "="*80 + "\n")

#E agora, fazer um gráfico
plt.figure(figsize=(10, 5))
contagens.head(10).plot(kind='barh')    #gráfico horizontal das 10 principais
plt.title(f'Top 10 valores em {coluna_para_grafico}')
plt.xlabel('Quantidade')
plt.ylabel(coluna_para_grafico)
plt.tight_layout()
plt.show()
```

```python
# CÉLULA - EXERCÍCIO
# Entendendo MÉDIA X MEDIANA

print("EXPLICANDO A TABELA:\n")

print("=" * 80)
print("SALÁRIOS - EXPLICAÇÃO")
print("=" * 80)

media = df['Salary_USD'].mean()
mediana = df['Salary_USD'].median()
minimo = df['Salary_USD'].min()
maximo = df['Salary_USD'].max()
desvio = df['Salary_USD'].std()

print(f"\nMÉDIA (soma de todos % quantidade): ${media:,.2f}")
print(f"\nMEDIANA (o valor do meio - sem sofrer com outliers): ${mediana:,.2f}")
print(f"\nMÍNIMO (o menor salário): ${minimo:,.2f}")
print(f"\nMÁXIMO (o maior salário): ${maximo:,.2f}")
print(f"\nDESVIO PADRÃO (Variação salarial): ${desvio:,.2f}")

print(f"\nSignifica dizer que:")
print(f"  - Salário típico: ${mediana:,.2f}")
print(f"  - Média puxada por salários altos: ${media:,.2f}")
print(f"  - Diferença: ${media - mediana:,.2f}")
print(f"  - Variação (±1 desvio): ${mediana - desvio:,.2f} até ${mediana + desvio:,.2f}")

print("\n" + "=" * 80)
print("QUARTIS (dividindo em 4 partes iguais)")
print("=" * 80)

q1 = df['Salary_USD'].quantile(0.25)
q2 = df['Salary_USD'].quantile(0.50)
q3 = df['Salary_USD'].quantile(0.75)

print(f"\n25% ganham menos que: ${q1:,.2f}")
print(f"\n50% ganham menos que: ${q2:,.2f} (mediana)")
print(f"\n75% ganham menos que: ${q3:,.2f}")
print(f"\nOu seja:")
print(f"  - 25% dos piores pagos: até ${q1:,.2f}")
print(f"  - 50% no meio: ${q1:,.2f} até ${q3:,.2f}")
print(f"  - 25% dos melhores pagos: acima de ${q3:,.2f}")
```
