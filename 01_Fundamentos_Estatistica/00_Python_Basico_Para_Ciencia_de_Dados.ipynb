```python
#Célula 10.1 : Explicação do Box Plot em detalhes
#Resolvi criar, porque a primeira vista, entender os bigodes, mediana.. o gráfico em si, é um tanto mais complicado para algumas pessoas, inclusive para mim!

# CÉLULA 10B: Explicar o Box Plot em Detalhes

print("="*80)
print("ENTENDENDO O BOX PLOT")
print("="*80)

# Calcular os valores exatos
Q1 = df['Salary_USD'].quantile(0.25)
Q2 = df['Salary_USD'].quantile(0.50)  # Mediana
Q3 = df['Salary_USD'].quantile(0.75)
IQR = Q3 - Q1  # Intervalo Interquartil
minimo = df['Salary_USD'].min()
maximo = df['Salary_USD'].max()

# Calcular os "bigodes" (whiskers)
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print(f"\n📊 ESTRUTURA DO BOX PLOT:\n")
print(f"Q1 (25%):          ${Q1:,.0f}")
print(f"  └─ 25% ganham MENOS que isso\n")

print(f"MEDIANA (50%):     ${Q2:,.0f}") 
print(f"  └─ A linha no MEIO da caixa é a mediana\n")

print(f"Q3 (75%):          ${Q3:,.0f}")
print(f"  └─ 75% ganham MENOS que isso\n")

print(f"IQR (Intervalo):   ${IQR:,.0f}")
print(f"  └─ A altura da caixa\n")

print(f"Limite inferior:   ${limite_inferior:,.0f}")
print(f"  └─ Ponto inicial dos 'bigodes'\n")

print(f"Limite superior:   ${limite_superior:,.0f}")
print(f"  └─ Ponto final dos 'bigodes'\n")

print(f"\nMínimo real:       ${minimo:,.0f}")
print(f"Máximo real:       ${maximo:,.0f}")

print("\n" + "="*80)
print("O QUE SIGNIFICA ISSO:")
print("="*80)

print(f"""
A CAIXA (do Q1 ao Q3):
  - Contém 50% dos dados
  - De ${Q1:,.0f} até ${Q3:,.0f}
  - A linha do meio = MEDIANA (${Q2:,.0f})

OS BIGODES:
  - Saem de Q1 e Q3
  - Vão até ${limite_inferior:,.0f} e ${limite_superior:,.0f}
  - Representam dados "normais"

OS PONTINHOS (se tiver):
  - Ficam ALÉM dos bigodes
  - São OUTLIERS (dados muito diferentes)
  - Neste caso: pessoas que ganham MUITO ou MUITO POUCO

SIMETRIA:
  - Se a mediana fica NO MEIO da caixa → simétrico
  - Se a mediana fica PARA UM LADO → enviesado
  - Neste caso, está {"Para baixo (assimétrico)" if Q2 < (Q1 + Q3) / 2 else "equilibrado"}
""")

print("="*80)
```

```python
#Após a explicação mais detalhada do box plot, resolvi criar mais uma, identificando outliers. Porque por mais que $8.080 e $233.977 sejam 'outliers' tecnicamente, o box plot não marca como 'ponto', porque estão dentro do limite calculado (Q3+1.5*IQR) - ou seja, são valores EXTREMOS, mas não "anormais"

# CÉLULA 10C: Identificar Outliers de Verdade
#E havendo dúvida, O que é IQR? Interquartile Range = Intervalo Interquartil - ou melhor, é a distância entre o terceiro quartil(Q3) e o primeiro quartil(Q1). É importante porque IQR ignora os extremos, por exemplo: suponhamos que houvessem salários: [100,200,300,400,500,1000000] o IQR, ignoraria '100' e '1000000'

print("="*80)
print("IDENTIFICANDO OUTLIERS")
print("="*80)

Q1 = df['Salary_USD'].quantile(0.25)
Q3 = df['Salary_USD'].quantile(0.75)
IQR = Q3 - Q1

# Limites para outliers
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print(f"\nQ1: ${Q1:,.0f}")
print(f"Q3: ${Q3:,.0f}")
print(f"IQR: ${IQR:,.0f}")
print(f"\nLimite inferior para outlier: ${limite_inferior:,.0f}")
print(f"Limite superior para outlier: ${limite_superior:,.0f}")

# Encontrar outliers
outliers = df[(df['Salary_USD'] < limite_inferior) | (df['Salary_USD'] > limite_superior)]

print(f"\n{'='*80}")
print(f"RESULTADO: {len(outliers)} OUTLIERS ENCONTRADOS")
print(f"{'='*80}")

if len(outliers) > 0:
    print(f"\nSalários abaixo de ${limite_inferior:,.0f}:")
    abaixo = df[df['Salary_USD'] < limite_inferior]['Salary_USD'].sort_values()
    if len(abaixo) > 0:
        print(f"  {len(abaixo)} pessoas")
        print(f"  Exemplos: {abaixo.head(3).values}")
    else:
        print("  Nenhum")
    
    print(f"\nSalários acima de ${limite_superior:,.0f}:")
    acima = df[df['Salary_USD'] > limite_superior]['Salary_USD'].sort_values(ascending=False)
    if len(acima) > 0:
        print(f"  {len(acima)} pessoas")
        print(f"  Exemplos: {acima.head(3).values}")
    else:
        print("  Nenhum")
else:
    print("\nNenhum outlier estatístico detectado!")
    print(f"\nMesmo que haja diferença entre ${df['Salary_USD'].min():,.0f} e ${df['Salary_USD'].max():,.0f},")
    print("estatisticamente não são considerados 'outliers' segundo o método IQR.")

print(f"\n{'='*80}")
print("CONCLUSÃO:")
print(f"{'='*80}")
print("""
Um valor é "outlier estatístico" quando está FORA do intervalo:
[Q1 - 1.5*IQR, Q3 + 1.5*IQR]

Mesmo que a diferença visual seja grande (8.080 vs 233.977),
se estiver dentro desse intervalo, não é outlier pelo critério IQR.

Isso não significa que o valor não seja INTERESSANTE.
Significa que está dentro da VARIAÇÃO ESPERADA dos dados.
""")
```

```python
#Célula 12: Gráficos das colunas categóricas

import matplotlib.pyplot as plt

# Criar figura com vários gráficos
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Distribuição das Colunas Categóricas', fontsize=16, fontweight='bold', y=1.00)

# Achatar o array de axes para ficar mais fácil iterar
axes = axes.flatten()

# Definir as colunas categóricas
colunas_cat = ['Job_Title', 'Country', 'Company_Type', 'Experience_Level', 'Remote', 'Top_Skill']

# Criar um gráfico para cada coluna
for idx, coluna in enumerate(colunas_cat):
    contagem = df[coluna].value_counts()

    # Gráfico de barras horizontal
    axes[idx].barh(contagem.index, contagem.values, color='steelblue', edgecolor='black', alpha=0.7)
    axes[idx].set_xlabel('Quantidade', fontsize=10)
    axes[idx].set_title(f'{coluna}', fontsize=11, fontweight='bold')
    axes[idx].grid(axis='x', alpha=0.3)

    # Adicionar números nas barras
    for i, v in enumerate(contagem.values):
        percentual = (v / len(df)) * 100
        axes[idx].text(v + 5, i, f'{v} ({percentual:.1f}%)', va='center', fontsize=9)

plt.tight_layout()
plt.show()

print("\n" + "="*80)
print("ANÁLISE VISUAL DAS CATEGORIAS")
print("\n" + "="*80)
```

```python
#Célula 13: Salário Médio por Categoria

print("="*80)
print("SALÁRIO MÉDIO POR CATEGORIA")
print("="*80)

# Definir as colunas categóricas
colunas_cat = ['Job_Title', 'Country', 'Company_Type', 'Experience_Level', 'Remote', 'Top_Skill']

for coluna in colunas_cat:
    print(f"\n\n{'='*80}")
    print(f"{coluna}")
    print(f"{'='*80}\n")

    # Agrupar por categoria e calcular estatísticas de salário
    grupo = df.groupby(coluna)['Salary_USD'].agg([
    ('Média', 'mean'),
    ('Mediana', 'median'),
    ('Min', 'min'),
    ('Max', 'max'),
    ('Quantidade', 'count')
    ]).round(2)

    # Ordenar por Média (do maior para o menor)
    grupo = grupo.sort_values('Média', ascending=False)

    print(grupo.to_string())
```

```python
# CÉLULA 14: Visualizar Salário Médio por Categoria

import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Salário Médio por Categoria', fontsize=16, fontweight='bold', y=1.00)

axes = axes.flatten()

colunas_cat = ['Job_Title', 'Country', 'Company_Type', 'Experience_Level', 'Remote', 'Top_Skill']

for idx, coluna in enumerate(colunas_cat):
    # Calcular média de salário por categoria
    media_por_categoria = df.groupby(coluna)['Salary_USD'].mean().sort_values(ascending=False)
    
    # Cores: verde para maior, vermelho para menor
    cores = plt.cm.RdYlGn([(x - media_por_categoria.min()) / (media_por_categoria.max() - media_por_categoria.min()) 
                            for x in media_por_categoria.values])
    
    # Gráfico de barras
    axes[idx].bar(range(len(media_por_categoria)), media_por_categoria.values, color=cores, edgecolor='black', alpha=0.8)
    axes[idx].set_xticks(range(len(media_por_categoria)))
    axes[idx].set_xticklabels(media_por_categoria.index, rotation=45, ha='right', fontsize=9)
    axes[idx].set_ylabel('Salário (USD)', fontsize=10)
    axes[idx].set_title(f'Salário Médio por {coluna}', fontsize=11, fontweight='bold')
    axes[idx].grid(axis='y', alpha=0.3)
    
    # Adicionar valores nas barras
    for i, v in enumerate(media_por_categoria.values):
        axes[idx].text(i, v + 1000, f'${v:,.0f}', ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()

print("\n" + "="*80)
print("INSIGHTS: QUAL CATEGORIA PAGA MAIS?")
print("="*80)

for coluna in colunas_cat:
    maior = df.groupby(coluna)['Salary_USD'].mean().sort_values(ascending=False).iloc[0]
    menor = df.groupby(coluna)['Salary_USD'].mean().sort_values(ascending=False).iloc[-1]
    diff = maior - menor
    percentual = (diff / menor) * 100
    
    categoria_maior = df.groupby(coluna)['Salary_USD'].mean().sort_values(ascending=False).index[0]
    categoria_menor = df.groupby(coluna)['Salary_USD'].mean().sort_values(ascending=False).index[-1]
    
    print(f"\n{coluna}:")
    print(f"  Maior: {categoria_maior} (${maior:,.0f})")
    print(f"  Menor: {categoria_menor} (${menor:,.0f})")
    print(f"  Diferença: ${diff:,.0f} ({percentual:.1f}%)")
```
